```python
# ---VARIABLES & TYPES ---
x = 5           #integer (whole number)
y = 3.14        # a float (decimal number)
name = "CGPA"   # a string text
is_placed = True
print(x, type(x))
print(y, type(y))
print(name, type(name))
print(is_placed, type(is_placed))
```

```python
# --- ARITHMEITC OPERATIONS ---
a = 7
b = 2
print("a + b =", a + b)
print("a - b =", a - b)
print("a * b =", a * b)
print("a / b =", a / b)
print("a ** b =", a ** 2)
# --- COMPARISION OPERATORS ---
cgpa = 8.4
print("cgpa == 8.4 :", cgpa == 8.4)
print("cgpa > 8     :", cgpa > 8)
print("cgpa < 6      :",cgpa < 6)
# --- LOGICAL OPERATORS ---
attendance = 85
placed = (cgpa > 8) and (attendance > 80)
print("placed :", placed)
```
